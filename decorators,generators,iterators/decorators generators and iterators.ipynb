```python
#DECORATORS:
def wraper(func):
    def inner_function(ls):
        total_sum = sum(ls)
        count = len(ls)
        return func(total_sum, count)
    return inner_function

@wraper
def average_finder(total_sum, count):
    print(f"Your average is: {round(total_sum / count, 2)}")

# Now, call average_finder with the list of numbers
ls = [25, 41, 63, 96, 85, 7]
average_finder(ls)
```

```python
def average(totalaverage):
    def insidefunction(ls):
        tsum=sum(ls)
        count=len(ls)
        totalaverage(tsum,count)
    return insidefunction

@average
def totalaverage(tsum,count):
    print(f"your average of the list is:{(tsum/count)}")
ls=[1,1,2,3,3,4,5,7]
totalaverage(ls)




```

```python
def greet_decorator(func):
    def wrapper():
        print("Hello, welcome!")
        func()
    return wrapper

@greet_decorator
def say_hello():
    print("This is the ANSHIKA GAUTAM")

say_hello()
```

```python
def simple_decorator(func):
    def wrapper(name):
        print("Start")
        func(name)
        print("End")
    return wrapper

@simple_decorator
def greet(name):
    for i in name:
     print(i)
    print(f"Hello, {name}!")

greet("Alice")
```

### GENERATORS

```python
def test(n):
    ls=[]
    for i in range(n):
        ls.append(i)
    return ls
# test(10)
print(test(10))
```

```python
def test(n):
    for i in range(n):
        yield i
```

```python
test(10)
```

```python
a=test(10)
print(a)
```

```python
for i in a:
    print(i)
```

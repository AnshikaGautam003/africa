### read and write in CSV file

```python
import csv
with open('output.csv', mode='r') as file:

## reader is used for reading the file instead of using read.
    csv_reader = csv.reader(file)
    for row in csv_reader:
        print(row)

```

```python
import csv

# Writing to a CSV file with writer() instead of using write().
data = [['Name', 'Age', 'City'], ['Alice', 25, 'New York'], ['Bob', 30, 'London']]

with open('output.csv', mode='w', newline='') as file:
    csv_writer = csv.writer(file)
    csv_writer.writerows(data)
```

### read and write in json file

```python
import json

# Reading JSON file with using load() instead of read
with open('output.json', mode='r') as file:
    data = json.load(file)
    print(data)
```

```python
import json

# Writing to a JSON file by using dump() instead of using of write.
data = {'Name': 'Alice', 'Age': 25, 'City': 'New York'}

with open('output.json', mode='w') as file:
    json.dump(data, file, indent=4)
```

```python
symbols=string.digits + string.ascii_letters
```

```python
symbols
```

```python
list(symbols)
```

```python
import random
```

```python
unique_id=random.choices(list(symbols),k=8)
```

```python
print("".join(unique_id))
```

```python
def downloader(images):
    for i in range(images):
        unique_id = random.choices(list(symbols),k=8)
        url = "https://picsum.photos/2000/3000"
        response = requests.get(url)
        with open ('image'+''.join(unique_id)+".jpg",'wb')as file:
            file.write(response.content)
```

```python
t1=Thread(target=downloader,args=[4])
t2=Thread(target=downloader,args=[6])
```

```python
t1.start()
t2.start()
```
